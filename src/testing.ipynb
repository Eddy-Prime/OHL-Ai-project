```python

base_path = '~/Desktop/SCHOOL/SEM2/Advanced AI/BUSit week/Attendance AI/Project 2 - Matchday Attendance Prediction/Data/'
df_match = pd.read_csv(base_path + 'gold_match.csv')
df_tickets = pd.read_csv(base_path + 'gold_match_tickets.csv')
df_context = pd.read_csv(base_path + 'gold_match_context.csv')


merged = df_match[['match_id', 'tickets_scanned']].merge(
    df_tickets[['match_id', 'tickets_sold_total']], on='match_id', how='inner'
)

merged['difference'] = merged['tickets_sold_total'] - merged['tickets_scanned']
merged['diff_pct'] = (merged['difference'] / merged['tickets_sold_total'] * 100).round(2)

print(merged[['match_id', 'tickets_sold_total', 'tickets_scanned', 'difference', 'diff_pct']])
print(f"\nAvg difference (sold - scanned): {merged['difference'].mean():.2f}")
print(f"Avg % not scanned: {merged['diff_pct'].mean():.2f}%")
```
